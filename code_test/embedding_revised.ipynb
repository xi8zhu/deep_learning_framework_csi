```python
import torch
from torch import nn
```

```python
linear = nn.Linear(32 * 2, 512)
x = torch.randn(50,4,64)
x.shape
```

```python
y = linear(x)
y.shape
```

```python
lstm = nn.LSTM(input_size = 512, hidden_size = 64, batch_first = True)
output, (h_n, c_n) = lstm(y)
output.shape
```
